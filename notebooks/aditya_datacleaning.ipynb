```python
df_required_column = df[['Utilities','Street','CentralAir','Electrical','BedroomAbvGr',
                         'KitchenAbvGr','PoolArea', 'GrLivArea', 'TotalBsmtSF', 
                         'GarageArea', '1stFlrSF', 'FullBath', 'GarageCars', 
                         'Neighborhood', 'MSZoning', 'YearBuilt', 'YearRemodAdd', 
                         'YrSold', 'MoSold', 'SaleCondition', 'OverallCond', 'ExterQual','OverallQual','SalePrice']].copy()
```

```python
df_required_column['Electrical'].unique()
```

```python
df_required_column['Electrical'].value_counts()
```

```python
df_required_column['Electrical'].fillna('SBrkr',inplace=True)
df_required_column['Electrical'].value_counts()
```

```python
df_required_column.isna().sum()
```

```python
df_required_column['Utilities'] = df_required_column.Utilities.replace({'AllPub': "All_Public_Facilities",
                                                                        "NoSeWa":"Only_Electricity_gas"})
```

```python
df_required_column['Street'] = df_required_column.Street.replace({'Pave': "Paved Road","Grvl":"Gravel Road"})
```

```python
df_required_column['Electrical'] = df_required_column.Electrical.replace({"SBrkr":"Standard_Circiut_Breaker",
                                                                          "FuseA":"Fuse_Good_Condition",
                                                                          "FuseF":"Fuse_Fair_Condition",
                                                                          "FuseP":"Fuse_poor_Condition",
                                                                          "mix":"Mixed_type"})
```

```python
df_required_column['ExterQual'] = df_required_column.ExterQual.replace({"Gd":"Good","TA":"Average","Ex":"Excellent","Fa":"Fair"})
```

```python
df_required_column['Neighborhood'] = df_required_column['Neighborhood'].astype(str).str.strip().str.title()
```
